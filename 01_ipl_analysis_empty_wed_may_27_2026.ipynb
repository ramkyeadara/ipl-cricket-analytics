```python
from pyspark.sql.functions import sum, count, round, col

# Top 10 batters by strike rate (2021 onwards, min 200 balls)
strike_rates = df.filter(
        (col("is_legal") == True) & 
        (col("season").isin("2025", "2026"))
    ) \
    .groupBy("striker") \
    .agg(
        sum("runs_off_bat").alias("runs"),
        count("runs_off_bat").alias("balls")
    ) \
    .withColumn("strike_rate", round(col("runs") * 100 / col("balls"), 2)) \
    .filter(col("balls") >= 200) \
    .orderBy(col("strike_rate").desc()) \
    .limit(10)

strike_rates.show()
```
